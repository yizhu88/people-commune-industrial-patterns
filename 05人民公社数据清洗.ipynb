```python
import pandas as pd

# 读取Excel文件
file_path = r"E:\【论文】\heritage science\people's commune0609.csv"
df = pd.read_csv(file_path)
```

```python
df['经纬度'] = df['经度'].astype(str) + ',' + df['纬度'].astype(str)
```

```python
df.head()
```

```python
columns_to_drop = ['经度', '纬度']
df.drop(columns=columns_to_drop, inplace=True)
```

```python
df.head()
```

```python
# 定义一个函数来将数据分为五级
def classify_column(series):
    # 根据数据的值的分布，使用分位数进行分类
    bins = [series.min(), series.quantile(0.2), series.quantile(0.4), series.quantile(0.6), series.quantile(0.8), series.max()]
    labels = [1, 2, 3, 4, 5]
    return pd.cut(series, bins=bins, labels=labels, include_lowest=True)

# 指定需要重新分级的列名
columns_to_classify = ['海拔', '降雨', '温度', '风力', '距运河', '距铁路', '距海洋', '距河流','距公路']

# 对每个列进行分级并创建新列
for col in columns_to_classify:
    df[col + '_level'] = classify_column(df[col])

# 显示处理后的数据，使用Pandas的标准显示方法
df.head()  # 显示前几行数据，你可以调整查看更多行
```

```python
# 删除原来的列
columns_to_drop = ['海拔', '降雨', '温度', '风力', '距运河', '距铁路', '距海洋', '距河流','距公路']
df.drop(columns=columns_to_drop, inplace=True)
```

```python
import pandas as pd

# 读取 CSV 文件
df = pd.read_csv(r"E:\【论文】\heritage science\0609Chinese_knowledge_graph\0609processed_data.csv")
```

```python
# 创建一个空的列表，用于存储三元组数据
triples = []

# 遍历 DataFrame 的每一行
for index, row in df.iterrows():
    head = row['COL_C']
    location = row['经纬度']
    
    # 需要转化为关系的列（不包括head和location）
    relations = ['专区', '市县', '现今', '海拔', '降雨', '气温', '风力', '至运河距离', '至铁路距离', '至海洋距离', '至河流距离', '至公路距离']
    
    # 遍历关系列并构造三元组
    for relation in relations:
        tail = row[relation]
        triples.append([head, tail, relation, location])

# 将三元组数据转化为 DataFrame
triples_df = pd.DataFrame(triples, columns=['head', 'tail', 'relation', 'location'])
```

```python
# 保存为新的 CSV 文件
triples_df.to_csv(r"E:\【论文】\heritage science\0609Chinese_knowledge_graph\0609三元组.csv", index=False)

# 显示部分结果
print(triples_df.head())
```

```python
import pandas as pd

# 读取含 location 的 CSV 文件
df_location = pd.read_csv(r"E:\【论文】\heritage science\0609Chinese_knowledge_graph\0609三元组.csv")

# 读取缺少 location 的 Excel 文件
df_new = pd.read_excel(r"E:\people's commune_classified.xlsx")
```

```python
# 提取需要的列
df_location = df_location[['head', 'tail', 'relation', 'location']].drop_duplicates()
df_new = df_new[['head', 'tail', 'relation']]  # 此文件无 location 列

# 构建 head → location 映射
head_location_mapping = df_location[['head', 'location']].drop_duplicates()

# 合并：按 head 补充 location
df_merged = pd.merge(df_new, head_location_mapping, on='head', how='left')

# 重新排列列顺序为 head, tail, relation, location
df_merged = df_merged[['head', 'tail', 'relation', 'location']]
```

```python
# 保存结果为新文件（可选）
df_merged.to_csv(r"E:\【论文】\heritage science\0609Chinese_knowledge_graph\0609三元组（含经纬度、产业）.csv", index=False, encoding='utf-8-sig')

# 显示部分结果以验证
df_merged.head()
```
