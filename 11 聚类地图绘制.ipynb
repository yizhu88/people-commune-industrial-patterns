```python
# -----------------------------
# 1. 导入库
# -----------------------------
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------------
# 2. 设置文件路径
# -----------------------------
file_path = r"E:\00论文\heritage science\0630\Extract_commune_with_cluster.csv"

# -----------------------------
# 3. 读取数据
# -----------------------------
df = pd.read_csv(file_path, encoding='utf-8')  # 如出错尝试 encoding='gbk'

# -----------------------------
# 4. 设置英文类别名称
# -----------------------------
cluster_labels = {
    0: "Traditional Agriculture",
    1: "Fishery & Sideline Economy",
    2: "Agro-Industrial Integration"
}
df['Cluster Type'] = df['cluster'].map(cluster_labels)

# -----------------------------
# 5. 绘图设置
# -----------------------------
plt.figure(figsize=(10, 8))
sns.scatterplot(
    data=df,
    x='Longitude',
    y='Latitude',
    hue='Cluster Type',
    palette='Set1',
    s=60,
    edgecolor='black'
)

plt.title("Geographic Distribution of People's Commune Types", fontsize=14)
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.legend(title="Commune Type")
plt.grid(True)
plt.tight_layout()
plt.show()
```
